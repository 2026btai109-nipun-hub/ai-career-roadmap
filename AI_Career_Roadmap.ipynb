### Step 2: Define Core AI Roadmap Engine

````python
import os
import json
import pandas as pd
from google import genai

# Enter your Gemini API key from https://aistudio.google.com/
GEMINI_API_KEY = "YOUR_GEMINI_API_KEY"

def generate_career_roadmap(current_skills, target_role, hours_per_week):
    client = genai.Client(api_key=GEMINI_API_KEY)
    
    prompt = f"""
    You are a senior tech career mentor. Compare the student's current skills against the target job role,
    identify the skill gap, and build a tailored 8-week learning plan based on their available study time.
    
    Student Profile:
    - Current Skills: {current_skills}
    - Target Job Role: {target_role}
    - Available Study Time: {hours_per_week} hours/week (Total: {hours_per_week * 8} hours)
    
    Return ONLY valid JSON matching this schema:
    {{
      "readiness_score": 55,
      "executive_summary": "...",
      "skill_gap_analysis": [ {{"domain_category": "...", "skills_you_have": "...", "skills_to_learn": "...", "priority": "High|Medium|Low"}} ],
      "eight_week_roadmap": [ {{"week_number": 1, "theme": "...", "key_topics": ["..."], "weekly_milestone": "..."}} ],
      "portfolio_projects": [ {{"title": "...", "description": "...", "tech_stack": ["..."], "resume_impact": "..."}} ]
    }}
    """
    
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )
    
    raw = response.text.strip().replace("```json", "").replace("```", "").strip()
    return json.loads(raw)
````

### Step 3: Run Example Test (Input ➔ AI ➔ Output)

````python
# Example Inputs
current_skills = "C, basic Python, HTML, basic SQL, Excel"
target_role = "Data Scientist"
hours_per_week = 12

print(f"Analyzing roadmap for: {target_role} ({hours_per_week} hrs/week)...")
result = generate_career_roadmap(current_skills, target_role, hours_per_week)

print(f"\n=== Starting Readiness: {result['readiness_score']}% ===")
print("Advisor Note:", result['executive_summary'])

print("\n--- 1. Skill Gap Analysis Table ---")
df = pd.DataFrame(result['skill_gap_analysis'])
display(df)

print("\n--- 2. 8-Week Learning Roadmap ---")
for w in result['eight_week_roadmap']:
    print(f"Week {w['week_number']}: {w['theme']}")
    print(f"  Topics: {', '.join(w['key_topics'])}")
    print(f"  Milestone: {w['weekly_milestone']}\n")

print("--- 3. Portfolio Projects ---")
for p in result['portfolio_projects']:
    print(f"* {p['title']}: {p['description']}")
    print(f"  Tech Stack: {', '.join(p['tech_stack'])}")
    print(f"  Impact: {p['resume_impact']}\n")
````
